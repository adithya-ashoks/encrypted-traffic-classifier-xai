```python
import pandas as pd

df = pd.read_csv('../data/raw/consolidated_traffic_data.csv')
print(df.shape)
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

# Separate features (X) and target (y)
X = df.drop('traffic_type', axis=1)
y = df['traffic_type']

# Encode labels to integers
le = LabelEncoder()
y_encoded = le.fit_transform(y)

# See the mapping (useful for interpreting results later)
label_mapping = dict(zip(le.classes_, [int(x) for x in le.transform(le.classes_)]))
print(label_mapping)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded,
    test_size=0.2,       # 80% train, 20% test
    random_state=42,     # ensures reproducibility - same split every time you run this
    stratify=y_encoded   # preserves class proportions in both sets
)

print("Train shape:", X_train.shape)
print("Test shape:", X_test.shape)
```

```python
import numpy as np

# Save as CSV (features) and numpy arrays (labels)
X_train.to_csv('../data/processed/X_train.csv', index=False)
X_test.to_csv('../data/processed/X_test.csv', index=False)
np.save('../data/processed/y_train.npy', y_train)
np.save('../data/processed/y_test.npy', y_test)

# Also save the label mapping so everyone can decode predictions back to class names
import json
with open('../data/processed/label_mapping.json', 'w') as f:
    json.dump(label_mapping, f, indent=2)

print("Saved successfully.")
```
